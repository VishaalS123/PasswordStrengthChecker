A simple cybersecurity project that checks password strength using five basic rules. It checks the password locally and does not store or transmit it.

Run the code cell, then enter a test password when prompted.

```python
"""
Project: Password Strength Checker
Student: Vishaal S

A simple educational tool that checks password strength locally.
It does not save or transmit the password.
"""

import re
import getpass

def check_password_strength(password):
    """Return a strength label and feedback for a password."""
    score = 0
    feedback = []

    if len(password) >= 8:
        score += 1
    else:
        feedback.append("Use at least 8 characters.")

    if re.search(r"[A-Z]", password):
        score += 1
    else:
        feedback.append("Add an uppercase letter (A-Z).")

    if re.search(r"[a-z]", password):
        score += 1
    else:
        feedback.append("Add a lowercase letter (a-z).")

    if re.search(r"\d", password):
        score += 1
    else:
        feedback.append("Add a number (0-9).")

    if re.search(r"[^A-Za-z0-9]", password):
        score += 1
    else:
        feedback.append("Add a special character (for example, ! or #).")

    if score <= 2:
        strength = "Weak"
    elif score <= 4:
        strength = "Moderate"
    else:
        strength = "Strong"

    return strength, feedback

def main():
    print("=" * 40)
    print("       PASSWORD STRENGTH CHECKER")
    print("=" * 40)
    print("Your password is checked locally and is not stored.")
    password = getpass.getpass("Enter a password to check: ")

    strength, feedback = check_password_strength(password)
    print(f"\nPassword strength: {strength}")

    if feedback:
        print("Suggestions:")
        for item in feedback:
            print("- " + item)
    else:
        print("Good job! Your password meets all five basic checks.")

if __name__ == "__main__":
    main()
```
